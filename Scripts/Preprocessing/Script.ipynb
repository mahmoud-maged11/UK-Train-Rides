### Convert time to category

```python
f = pd.DataFrame({
    'Departure Time': ['04:30', '08:15', '12:45', '16:00', '20:30', '23:15', '02:10']
})

f['Departure Time'] = pd.to_datetime(f['Departure Time'], format='%H:%M')

def get_time_period(time):
    hour = time.hour
    if 4 <= hour < 6:
        return 'Dawn'
    elif 6 <= hour < 12:
        return 'Morning'
    elif 12 <= hour < 17:
        return 'Noon'
    elif 17 <= hour < 21:
        return 'After Noon'
    elif 21 <= hour < 24:
        return 'Night'
    elif 0 <= hour < 4:
        return 'Late Night'
    else:
        return 'Unknown'

data['TimePeriod'] = f['Departure Time'].apply(get_time_period)

data.head()
```
